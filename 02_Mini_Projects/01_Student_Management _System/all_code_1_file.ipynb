# All code of student Management system
 

```python
# ==========================================
#      STUDENT MANAGEMENT SYSTEM
# ==========================================

import json

students = []


# ==========================================
# Load Students
# ==========================================
def load_students():
    global students

    try:
        with open("students.json", "r") as file:
            students = json.load(file)

    except FileNotFoundError:
        students = []


# ==========================================
# Save Students
# ==========================================
def save_students():

    with open("students.json", "w") as file:
        json.dump(students, file, indent=4)


# ==========================================
# Add Student
# ==========================================
def add_student():

    try:

        student_id = int(input("Enter Student ID: "))

        # Check Duplicate ID
        for student in students:
            if student["id"] == student_id:
                print("Student ID already exists!")
                return

        name = input("Enter Name: ")

        age = int(input("Enter Age: "))

        student_class = input("Enter Class (BSCS/BSIT): ")

        marks = int(input("Enter Marks: "))

        student = {
            "id": student_id,
            "name": name,
            "age": age,
            "class": student_class,
            "marks": marks
        }

        students.append(student)

        save_students()

        print("\nStudent Added Successfully!")

    except ValueError:
        print("Please enter valid numeric values.")


# ==========================================
# View Students
# ==========================================
def view_students():

    if len(students) == 0:
        print("\nNo Students Found.")
        return

    print("\n========== STUDENTS ==========")

    for student in students:

        print("------------------------------")
        print("ID    :", student["id"])
        print("Name  :", student["name"])
        print("Age   :", student["age"])
        print("Class :", student["class"])
        print("Marks :", student["marks"])

    print("------------------------------")
    print("Total Students:", len(students))


# ==========================================
# Search Student
# ==========================================
def search_student():

    print("\nSearch By")

    print("1. ID")
    print("2. Name")

    choice = input("Enter Choice: ")

    if choice == "1":

        try:
            search_id = int(input("Enter Student ID: "))

        except ValueError:
            print("Invalid ID")
            return

        for student in students:

            if student["id"] == search_id:

                print("\nStudent Found")
                print("---------------------")
                print("ID    :", student["id"])
                print("Name  :", student["name"])
                print("Age   :", student["age"])
                print("Class :", student["class"])
                print("Marks :", student["marks"])

                return

        print("Student Not Found.")

    elif choice == "2":

        search_name = input("Enter Student Name: ").lower()

        for student in students:

            if student["name"].lower() == search_name:

                print("\nStudent Found")
                print("---------------------")
                print("ID    :", student["id"])
                print("Name  :", student["name"])
                print("Age   :", student["age"])
                print("Class :", student["class"])
                print("Marks :", student["marks"])

                return

        print("Student Not Found.")

    else:
        print("Invalid Choice.")


# ==========================================
# Update Student
# ==========================================
def update_student():

    try:

        search_id = int(input("Enter Student ID: "))

    except ValueError:
        print("Invalid ID")
        return

    for student in students:

        if student["id"] == search_id:

            print("\nStudent Found")

            student["name"] = input("New Name: ")

            student["age"] = int(input("New Age: "))

            student["class"] = input("New Class: ")

            student["marks"] = int(input("New Marks: "))

            save_students()

            print("Student Updated Successfully!")

            return

    print("Student Not Found.")


# ==========================================
# Delete Student
# ==========================================
def delete_student():

    try:

        search_id = int(input("Enter Student ID: "))

    except ValueError:
        print("Invalid ID")
        return

    for student in students:

        if student["id"] == search_id:

            confirm = input("Are you sure? (yes/no): ").lower()

            if confirm == "yes":

                students.remove(student)

                save_students()

                print("Student Deleted Successfully!")

            else:

                print("Deletion Cancelled.")

            return

    print("Student Not Found.")


# ==========================================
# Main Program
# ==========================================

load_students()

while True:

    print("\n===================================")
    print("      STUDENT MANAGEMENT SYSTEM")
    print("===================================")

    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Update Student")
    print("5. Delete Student")
    print("6. Exit")

    choice = input("\nEnter Choice: ")

    if choice == "1":

        add_student()

    elif choice == "2":

        view_students()

    elif choice == "3":

        search_student()

    elif choice == "4":

        update_student()

    elif choice == "5":

        delete_student()

    elif choice == "6":

        print("\nThank you for using Student Management System.")
        break

    else:

        print("Invalid Choice. Please try again.")
```
