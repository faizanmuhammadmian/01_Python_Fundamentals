#  File Handling in Python
File Handling is the process of creating, opening, reading, writing, updating, and deleting files using Python.

Instead of storing data only in variables (which disappear when the program ends), file handling allows you to save data permanently.

# Opening a File

Syntax:
```python
file = open("filename", "mode")
```
**Example**

````python
file = open("students.txt", "r")
````

## File Modes

| Mode   | Meaning           | File Must Exist? | Creates File?                |
| ------ | ----------------- | ---------------- | ---------------------------- |
| `"r"`  | Read              | ✅ Yes            | ❌ No                         |
| `"w"`  | Write (overwrite) | ❌ No             | ✅ Yes                        |
| `"a"`  | Append            | ❌ No             | ✅ Yes                        |
| `"x"`  | Create new file   | ❌ No             | ✅ Yes (fails if file exists) |
| `"r+"` | Read and Write    | ✅ Yes            | ❌ No                         |


## 1. Read a File (r)

Example:

````python
file = open("students.txt", "r")

print(file.read())

file.close()
````

#### Suppose `students.txt` contains:

Ali
Sara
Ahmed

# 2. Write to a File (`w`)

Creates a file or replaces all existing content.

````python
file = open("students.txt", "w")

file.write("Ali")

file.close()
````

# 3. Append to a File (a)

Adds data without deleting existing content.

````python
file = open("students.txt", "a")

file.write("\nSara")

file.close()
````

# 4. Create a New File (`x`)

````python
file = open("newfile.txt", "x")

file.close()
````

##### If `newfile.txt` already exists:

`FileExistsError`

## Closing a File

Always close a file after using it.
```python

file.close()
```

This frees system resources and ensures changes are saved.

# Using `with` :

Instead of manually calling `close()`, use `with`.



````python
with open("students.txt", "r") as file:
    print(file.read())
````



The file closes automatically when the block ends.

# Reading Line by Line

````python
with open("students.txt", "r") as file:

    for line in file:
        print(line.strip())
````

## `read()`

Reads the entire file.

````python
with open("students.txt", "r") as file:

    data = file.read()

print(data)
````

## `readline()`

Reads one line at a time.

````python
with open("students.txt", "r") as file:

    print(file.readline())
    print(file.readline())
````

## `readlines()`

Returns a list of lines.

````python
with open("students.txt", "r") as file:

    lines = file.readlines()

print(lines)
````

## File Pointer

Each time you read a file, the pointer moves forward.

````python
with open("students.txt", "r") as file:

    print(file.read())
    print(file.read())
````

The second read() prints nothing because the pointer is already at the end.

## Move the File Pointer

`seek(0)` moves the pointer back to the beginning.

````python
with open("students.txt", "r") as file:

    print(file.read())

    file.seek(0)

    print(file.read())
````

## Check if a File Exists

````python
import os

if os.path.exists("students.txt"):
    print("File Exists")
else:
    print("File Not Found")
````

Delete a File

````python
import os

os.remove("students.txt")
````
