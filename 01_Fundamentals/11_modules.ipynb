# Modules
A module is a Python file (.py) that contains functions, variables, classes, or code which can be reused in other Python programs.

Instead of writing the same code again and again, you write it once in a module and import it whenever you need it.


**Modules help us:**

- Reuse code
- Organize programs
- Reduce duplicate code
- Make projects easier to maintain
- Share code between projects

### Types of Modules

Python has three types of modules:



| Type                 | Example                            |
| -------------------- | ---------------------------------- |
| Built-in Modules     | `math`, `random`, `os`, `datetime` |
| User-defined Modules | Your own `.py` files               |
| Third-party Modules  | `numpy`, `pandas`, `matplotlib`    |


## 1. Built-in Module

Python already includes many useful modules.

**Example:**

````python
import math

print(math.sqrt(25))
print(math.pi)
print(math.factorial(5))
print(math.sqrt(64))
````

### random
Every time you run it, the number may be different.

````python
import random

print(random.randint(1,10))
````

### datetime

````python
import datetime

today = datetime.datetime.now()

print(today)
````

### `os`
(Current working directory)

````python
import os

print(os.getcwd())
````

## 2. User-Defined Module

Create your own module.

### Step 1

Create a file:
`calculator.py`

Write this code:

````python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b
````

### Step 2

Create another file:
`main.py`

````python
import calculator

print(calculator.add(10,5))
print(calculator.subtract(20,7))
````

## Import Specific Functions

Instead of importing the whole module:
```python
import calculator
```

Import only what you need.

````python
from calculator import add

print(add(5,3))
````

## Import Multiple Functions

````python

from calculator import add, subtract

print(add(4,6))
print(subtract(10,3))
````
