# Data Types

A data type tells Python what kind of value a variable stores.


| Data Type | Description                   | Example           |
| --------- | ----------------------------- | ----------------- |
| `int`     | Whole numbers                 | `20`              |
| `float`   | Decimal numbers               | `3.14`            |
| `str`     | Text                          | `"Ali"`           |
| `bool`    | True or False                 | `True`            |
| `list`    | Ordered, mutable collection   | `[1, 2, 3]`       |
| `tuple`   | Ordered, immutable collection | `(1, 2, 3)`       |
| `dict`    | Key-value pairs               | `{"name": "Ali"}` |
| `set`     | Unique values                 | `{1, 2, 3}`       |


```python
age = 20
marks = 95
print(f' age is {age} and its type is {type(age)}')
print(f' marks is {marks} and its type is {type(marks)}')

price = 99.99
height = 5.8
print(f' price is {price} and its type is {type(price)}')
print(f' height is {height} and its type is {type(height)}')

name = "Ali"
city = "Peshawar"
print(f' name is {name} and its type is {type(name)}')
print(f' city is {city} and its type is {type(city)}')

is_student = True
is_logged_in = False
print(f' is_student is {is_student} and its type is {type(is_student)}')
print(f' is_logged_in is {is_logged_in} and its type is {type(is_logged_in)}')
```
