# 4.0 Operators

An operator is a symbol that performs an operation on one or more values.

### Example:


> 5 + 3

- Here, + is an operator.




| Operator Type | Operators                           |                       
| ------------- | ----------------------------------- |  
| Arithmetic    | `+`, `-`, `*`, `/`, `//`, `%`, `**` |                         
| Comparison    | `==`, `!=`, `>`, `<`, `>=`, `<=`    |                         
| Assignment    | `=`, `+=`, `-=`, `*=`, `/=`, `%=`   |                         
| Logical       | `and`, `or`, `not`                  |                        
| Membership    | `in`, `not in`                      |                         
| Identity      | `is`, `is not`                      |
| Bitwise       | `&`, `, `^`, `~`, `<<`, `>>`        |                        



### 1. Arithmetic Operators
Used for mathematical calculations.


| Operator | Meaning             | Example   | Output |
| -------- | ------------------- | --------- | ------ |
| `+`      | Addition            | `5 + 3`   | `8`    |
| `-`      | Subtraction         | `5 - 3`   | `2`    |
| `*`      | Multiplication      | `5 * 3`   | `15`   |
| `/`      | Division            | `10 / 2`  | `5.0`  |
| `//`     | Floor Division      | `10 // 3` | `3`    |
| `%`      | Modulus (Remainder) | `10 % 3`  | `1`    |
| `**`     | Exponent (Power)    | `2 ** 3`  | `8`    |


````python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
````

### 2. Comparison (Relational) Operators :
Used to compare two values.


| Operator | Meaning                  | Example  | Output  |
| -------- | ------------------------ | -------- | ------- |
| `==`     | Equal to                 | `5 == 5` | `True`  |
| `!=`     | Not equal to             | `5 != 3` | `True`  |
| `>`      | Greater than             | `10 > 5` | `True`  |
| `<`      | Less than                | `5 < 10` | `True`  |
| `>=`     | Greater than or equal to | `5 >= 5` | `True`  |
| `<=`     | Less than or equal to    | `5 <= 4` | `False` |


````python
age = 20

print(age >= 18)
print(age < 30)
print(age == 20)
print(age != 25)
print(age <= 18 )
print(age < 40 )
````

### 3. Assignment Operators

Used to assign values to variables.

| Operator | Example  | Same As     |
| -------- | -------- | ----------- |
| `=`      | `x = 5`  | Assign 5    |
| `+=`     | `x += 2` | `x = x + 2` |
| `-=`     | `x -= 2` | `x = x - 2` |
| `*=`     | `x *= 2` | `x = x * 2` |
| `/=`     | `x /= 2` | `x = x / 2` |
| `%=`     | `x %= 2` | `x = x % 2` |


````python
x = 10
print(x)
x += 5
print(f'mean x=10,x=5 ,10+5={x}')
````

### 4. Logical Operators

Used to combine conditions.

| Operator | Meaning                                  |
| -------- | ---------------------------------------- |
| `and`    | Both conditions must be True             |
| `or`     | At least one condition must be True      |
| `not`    | Reverses True to False and False to True |


````python
age = 20
has_id = True

print(age >= 18 and has_id)
````

### 5. Membership Operators

Used to check whether a value exists in a sequence.


| Operator | Meaning              |
| -------- | -------------------- |
| `in`     | Value exists         |
| `not in` | Value does not exist |


````python
names = ["Ali", "Ahmed", "Sara"]

print("Ali" in names)
print("John" not in names)
````

### 6. Identity Operators

Used to check whether two variables refer to the same object.


| Operator | Meaning           |
| -------- | ----------------- |
| `is`     | Same object       |
| `is not` | Different objects |


````python
x = None

print(x is None)
````

### 7. Bitwise Operators (Advanced)

Used to work with binary (bits).

| Operator | Meaning     |    |
| -------- | ----------- | -- |
| `&`      | AND         |    |
| `        | `           | OR |
| `^`      | XOR         |    |
| `~`      | NOT         |    |
| `<<`     | Left Shift  |    |
| `>>`     | Right Shift |    |

#### 1. Bitwise AND (&)

Returns 1 only if both bits are 1.

Example:

```python
a = 5
b = 3
print(a & b)
```

Output:

`
1
`

Binary:

```
5 = 101
3 = 011
------------
    001 = 1
    ```

----------------------------------------------------------------



####  2. Bitwise OR (|)

Returns 1 if at least one bit is 1.

Example:
```python
a = 5
b = 3

print(a | b)
```

**Output:**

`
7
`

Binary:

``
5 = 101
3 = 011
------------
    111 = 7

#### 3. Bitwise XOR (^)

Returns 1 if the bits are different.

Example:

```python
a = 5
b = 3

print(a ^ b)
```

Output:

6

Binary:

5 = 101
3 = 011
------------
    110 = 6

#### 4. Bitwise NOT (~)

Reverses all bits.

Example:

```python 
a = 5

print(~a)
```

Output:

-6

Note: The result is negative because Python stores integers using a representation called two's complement. You don't need to understand the details yet.

#### 5. Left Shift (<<)

Shifts bits to the left.

Example:

``` python 
a = 5

print(a << 1)
```
Output:

10

Binary:

5  = 101

Shift left by 1

1010 = 10

A left shift by 1 is equivalent to multiplying by 2 (for positive integers).

#### 6. Right Shift (>>)

Shifts bits to the right.

Example:

```python
a = 8

print(a >> 1)
```

Output:

4

Binary:

8 = 1000

Shift right by 1

100 = 4

A right shift by 1 is roughly equivalent to dividing by 2 (for positive integers).
