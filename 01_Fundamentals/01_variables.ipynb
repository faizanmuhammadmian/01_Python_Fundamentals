# Variables

## Definition 

A variable is  a named container that store data in memory .

- Syntax:

`variable_name = value `

example :

- `` name  = "Faizan" ``
- `` age = 21  ``
- `` marks = 95 ``

# Rules of variables:
1. A variable name must start with a letter or an underscore (_).
2. A variable name cannot start with a number.
3. A variable name can contain only letters, numbers, and underscores (_).
4. A variable name cannot contain spaces.
5. Python is case-sensitive, so name, Name, and NAME are different variables.
6. A variable name cannot be a Python keyword (such as if, for, or class).
7. Use meaningful and descriptive variable names.
8. Follow Python's snake_case naming convention (e.g., student_name).
9. A variable's value can be changed by assigning it a new value.
10. A variable can store different types of data (such as integers, strings, floats, and booleans) during program execution
