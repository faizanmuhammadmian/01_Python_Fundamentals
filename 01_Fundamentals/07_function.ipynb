# Functions :

A function is a reusable block of code that performs a specific task. Instead of writing the same code multiple times, you write it once inside a function and call it whenever needed.

Syntax:
```python
def function_name():
    # Code
```
Example:

````python
def greet():
    print("Hello, World!")

greet()
````

- def → Keyword to define a function.
- greet → Function name.
- () → Parameters go here.
- : → Starts the function body.
- Indented code → Executes when the function is called.

## Parameters
Parameters let you pass data into a function.

````python
def greet(name):
    print("Hello", name)

greet("Ali")
````

## Return Statement

A function can return a value using return.

````python
def add(a, b):
    return a + b

result = add(10, 5)

print(result)
````

## Difference Between print() and return

Using print():

````python
def add(a, b):
    print(a + b)

add(2, 3)
#You cannot store the result.
````

You cannot store the result.

Using return:

````python
def add(a, b):
    return a + b

result = add(2, 3)
print(result)
````

Now the value can be stored and used later.

## Default Parameters

````python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Ahmed")
````

## Keyword Arguments

````python
def student(name, age):
    print(name)
    print(age)

student(age=21, name="Ali")
````

## Variable Scope
Local Variable

Exists only inside the function.

````python
def test():
    x = 10
    print(x)

test()
````

Outside the function:
```python
print(x)
```
This causes a NameError.

## Global Variable

Accessible everywhere.

````python


x = 100

def show():
    print(x)

show()
print(x)
````

## Lambda Function

A short anonymous function.

````python
square = lambda x: x * x

print(square(4))
````

## Recursive Function

A function that calls itself.

````python
def countdown(n):

    if n == 0:
        print("Done")
        return

    print(n)
    countdown(n - 1)

countdown(5)
````

## Built-in Functions

Python already provides many useful functions.

````python
print("Hello")

len("Python")

max(5, 10)

min(5, 10)

sum([1,2,3])

type(10)
````

## User-Defined Function

Functions created by the programmer.

````python
def multiply(a, b):
    return a * b

print(multiply(4, 5))
````

### Mini Project Example

````python
def calculate_grade(marks):

    if marks >= 90:
        return "A"

    elif marks >= 80:
        return "B"

    elif marks >= 70:
        return "C"

    elif marks >= 60:
        return "D"

    else:
        return "Fail"

score = int(input("Enter Marks: "))
print(f'You entered: {score}')
print("Grade:", calculate_grade(score))
````
