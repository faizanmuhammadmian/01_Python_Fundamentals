# Object-Oriented Programming (OOP):

Object-Oriented Programming (OOP) is a programming paradigm that organizes code into objects and classes. It helps make programs modular, reusable, organized, and easier to maintain.

> Without OOP, programs become difficult to manage as they grow.

OOP helps you:

- Reuse code
- Reduce duplication
- Organize large programs
- Improve readability
- Make maintenance easier


### Four Pillars of OOP
1. Class
2. Object
3. Inheritance
4. Polymorphism
5. Encapsulation
6. Abstraction

(The four main pillars are Inheritance, Polymorphism, Encapsulation, and Abstraction. Before learning them, you need to understand Class and Object.)

## 1. Class

A class is a blueprint or template for creating objects.

Syntax
class Student:
    pass

Here, Student is a class.

````python
class Car:
    pass
````

2. Object

An object is an instance of a class.

Example

````python
class Car:
    pass

car1 = Car()
car2 = Car()
````

Now:

- car1 is an object.
- car2 is another object.
Both are created from the same blueprint.

### Constructor (__init__)

The constructor runs automatically when an object is created.

Syntax:
```python
class Student:

    def __init__(self):
        print("Object Created")
```

Example:

````python
class Student:

    def __init__(self):
        print("Student Created")

student1 = Student()
````

#### The self Keyword

`self` refers to the current object.

Example:

````python
class Student:

    def __init__(self, name):
        self.name = name

student1 = Student("Ali")

print(student1.name)
````

#### Attributes

Attributes store data inside objects.

Example:

````python
class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age

student = Student("Sara", 20)

print(student.name)
print(student.age)
````

#### Methods

Methods are functions inside a class.

Example:

````python
class Student:

    def __init__(self, name):

        self.name = name

    def greet(self):
        print("Hello", self.name)

student = Student("Ahmed")

student.greet()
````

#### Complete Example

````python
class Student:

    def __init__(self, name, age, marks):

        self.name = name
        self.age = age
        self.marks = marks

    def display(self):

        print("Name :", self.name)
        print("Age :", self.age)
        print("Marks :", self.marks)

student1 = Student("Ali", 20, 90)

student1.display()
````

## Encapsulation

Encapsulation means keeping data and methods together inside a class and controlling access to the data.

Example

````python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def show_balance(self):
        print(self.__balance)

account = BankAccount(5000)

account.show_balance()
````

## Inheritance

Inheritance allows one class to inherit features from another class.

Example

````python
class Animal:

    def speak(self):
        print("Animal Sound")

class Dog(Animal):
    pass

dog = Dog()

dog.speak()
````

## Polymorphism

The same method behaves differently for different objects.

Example

````python
class Cat:

    def sound(self):
        print("Meow")

class Dog:

    def sound(self):
        print("Bark")

animals = [Cat(), Dog()]

for animal in animals:
    animal.sound()
````

## Abstraction

Abstraction hides unnecessary implementation details and shows only the essential features.

Example

````python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

class Square(Shape):

    def area(self):
        return 25

square = Square()

print(square.area())
````

### Access Modifiers  

| Modifier  | Syntax   | Access                             |
| --------- | -------- | ---------------------------------- |
| Public    | `name`   | Anywhere                           |
| Protected | `_name`  | Class & subclasses (by convention) |
| Private   | `__name` | Inside the class                   |
