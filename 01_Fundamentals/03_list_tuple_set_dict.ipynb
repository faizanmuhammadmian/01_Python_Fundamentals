# 1. Python list

- It  represent with square Breckets [ ].
- In list we can store different datatype.
- It allow duplicates value.

```python
a = [1,'love',2.7]  
print(a)          
print(type(a))

```


| Method                          | What it does                                                           |
| ------------------------------- | ---------------------------------------------------------------------- |
| `append(x)`                     | Adds `x` to the end of the list                                        |
| `extend(iterable)`              | Adds all elements of another iterable to the list                      |
| `insert(i, x)`                  | Inserts `x` at position `i`                                            |
| `remove(x)`                     | Removes the **first** occurrence of `x`                                |
| `pop([i])`                      | Removes and returns the item at index `i` (last item if `i` not given) |
| `clear()`                       | Removes all elements from the list                                     |
| `index(x[, start[, end]])`      | Returns index of first occurrence of `x`                               |
| `count(x)`                      | Returns how many times `x` appears                                     |
| `sort(key=None, reverse=False)` | Sorts the list in place                                                |
| `reverse()`                     | Reverses the list in place                                             |
| `copy()`                        | Returns a shallow copy of the list                                     |

#  4. Python dict   
- it  did not use  duplicates  keys



| Methods                            |What it does                                                                                           |
|:----------------------------------:|:-----------------------------------------------------------------------------------------------------:|  
| dict.clear()                       |Removes all items from the dictionary. e.g : d.clear()                                                 |
| dict.copy()                        |Returns a shallow copy of the dictionary. e.g : new_d = d.copy()                                       |
| dict.fromkeys(iterable, value=None)|Creates a new dictionary with given keys and a common value.e.g :d = dict.fromkeys(['a', 'b', 'c'], 0) |
| dict.get(key, default=None)        |Returns the value of a key; returns default if key doesn’t exist.e.g : d.get('age', 0)                 |
| dict.items()                       |Returns a view object of key–value pairs. e.g : d.items()                                              |
| dict.keys()                        | Returns a view object of all keys. e.g: d.keys()                                                      |
| dict.values()                      | Returns a view object of all values. e.g: d.values()                                                  |
| dict.pop(key, default)             | Removes and returns the value of the specified key. e.g: d.pop('name')                                |
| dict.popitem()                     | Removes and returns the last inserted key–value pair. e.g: d.popitem()                                |
| dict.setdefault(key, default=None) |Returns the value of a key; if not present, inserts it with a default value.e.g: d.setdefault('score', 0)|
| dict.update(other_dict)            | Updates the dictionary with elements from another dictionary or iterable.e.g: d.update({'age': 25})   |
