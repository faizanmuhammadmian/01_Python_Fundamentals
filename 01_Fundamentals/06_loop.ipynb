# Loops 
A loop is used to repeat a block of code multiple times until a condition is met




**Types of Loops in Python**

Python has two main loops:
- for Loop
- while Loop.

### 1. for Loop

A for loop is used when you know how many times you want to repeat something or when iterating over a collection.

Syntax:
```python
for variable in sequence:
    # Code
```
**Example 1**

````python
for i in range(5):
    print(i)

for i in range(2, 11, 2):
    print(i)    
````

#### Understanding range()

range() generates a sequence of numbers.

##### One Argument
```python
range(5)
```
Output:
```
0 1 2 3 4
```


##### Two Arguments

```python
range(1, 6)
```
Output:

```
1 2 3 4 5
```

##### Three Arguments
```
range(start, stop, step)
```
Example
```python
range(2, 11, 2)
```
Output
```
2 4 6 8 10
```

## while Loop

A while loop repeats as long as the condition is True.

Syntax:
```python
while condition:
    # Code
```    

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

````python
password = ""

while password != "python":
    password = input("Enter Password: ")

print("Access Granted")
# The loop keeps asking until the correct password is entered.
````

## Infinite Loop

```python
while True:
    print("Hello")
```

This loop never stops.

Stop it by pressing:
```
Ctrl + C
```
or by using a break statement.

## break Statement

Stops the loop immediately.

```python
for i in range(10):

    if i == 5:
        break

    print(i)
```
Output

```
0
1
2
3
4
```
## continue Statement

Skips the current iteration.

```python
for i in range(6):

    if i == 3:
        continue
    print(i)
```

Output
```
0
1
2
4
5
```
## pass Statement

Acts as a placeholder.
```python
for i in range(5):
    pass
```
Useful when writing code that you'll complete later.
