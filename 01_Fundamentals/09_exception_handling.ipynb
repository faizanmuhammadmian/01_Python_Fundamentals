# Exception Handling in Python

Exception Handling is a way to handle errors in a program so it doesn't crash unexpectedly. Instead of stopping, the program can display a meaningful message or recover gracefully.

## 1. try and except

The simplest form of exception handling.

Syntax:
```python
try:
    # Code that may cause an error
except:
    # Runs if an error occurs
```
**Example**:

````python
try:
    number = int(input("Enter a number: "))
    print(number)
except:
    print("Invalid input.")
````

### Common Errors (Exceptions)

| Exception           | Cause                  | Example            |
| ------------------- | ---------------------- | ------------------ |
| `ValueError`        | Invalid value          | `int("abc")`       |
| `TypeError`         | Wrong data type        | `"5" + 5`          |
| `ZeroDivisionError` | Division by zero       | `10 / 0`           |
| `IndexError`        | Invalid list index     | `numbers[10]`      |
| `KeyError`          | Missing dictionary key | `student["marks"]` |
| `FileNotFoundError` | File doesn't exist     | `open("data.txt")` |
| `NameError`         | Variable not defined   | `print(age)`       |


## 2. Catching Specific Exceptions

Instead of catching every error, catch only the one you expect.

Example: `ValueError`


````python
try:
    age = int(input("Enter Age: "))
    print(age)
except ValueError:
    print("Age must be a number.")
````

**Example**: `ZeroDivisionError`

````python
try:
    a = 10
    b = 0
    print(a / b)
except ZeroDivisionError:
    print("Cannot divide by zero.")
````

**Example**: `FileNotFoundError`

## 3. Multiple `except` Blocks

Handle different exceptions separately.

````python
try:
    number = int(input("Enter Number: "))
    print(10 / number)

except ValueError:
    print("Please enter numbers only.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
````

## 4. `else` Block

Runs only if no exception occurs.

````python
try:
    age = int(input("Enter Age: "))
except ValueError:
    print("Invalid Age")
else:
    print("Age is", age)
````

## 5. `finally` Block

Runs whether an exception occurs or not.

````python
try:
    file = open("students.txt")
except FileNotFoundError:
    print("File not found.")
finally:
    print("Program Finished")
````

## 6. Raising Exceptions

You can create your own exception using `raise`.

````python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative.")
````

## 7. Custom Exception

Create your own exception class.

````python
class InvalidAgeError(Exception):
    pass

age = -10

if age < 0:
    raise InvalidAgeError("Invalid Age")
````
